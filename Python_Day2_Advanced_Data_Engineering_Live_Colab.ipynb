## 6. Exception Handling
Use `try/except` to handle expected errors without crashing the whole program.


````python
try:
    value = int("abc")
except ValueError as error:
    print("Invalid integer:", error)
else:
    print("Conversion succeeded:", value)
finally:
    print("This always executes")
````

### Multiple Exception Types


````python
records = [{"amount":"100"}, {"amount":None}, {}, {"amount":"bad"}]

for record in records:
    try:
        amount = float(record["amount"])
        print("Processed:", amount)
    except (KeyError, TypeError, ValueError) as error:
        print("Rejected:", record, "|", type(error).__name__)
````

## 9. JSON
JSON represents structured/semi-structured data using objects and arrays. Python dictionaries map naturally to JSON objects.


````python
import json

customer = {
    "id": 101,
    "name": "John",
    "address": {"city": "London", "postcode": "E1 1AA"},
    "orders": [{"id": 1, "amount": 100}, {"id": 2, "amount": 250}]
}

with open("customer.json", "w") as file:
    json.dump(customer, file, indent=2)

with open("customer.json", "r") as file:
    data = json.load(file)

print(data["address"]["city"])
print(data["orders"][0]["amount"])
````

## 13. OOP Basics
A class is a blueprint; an object is an instance. `__init__` initializes object attributes.


````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print(f"{self.name}: £{self.salary}")

emp = Employee("John", 55000)
emp.display()
````

## 14. Encapsulation / Advanced OOP
OOP can group data and behavior and control how internal state is accessed.


````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount

    def get_balance(self):
        return self._balance

account = BankAccount("John", 1000)
account.deposit(500)
print(account.get_balance())
````

## 15. Inheritance
Inheritance lets a child class reuse and extend a parent class.


````python
class DataSource:
    def connect(self):
        print("Connecting to data source")

class DatabaseSource(DataSource):
    def query(self):
        print("Running SQL query")

db = DatabaseSource()
db.connect()
db.query()
````

## 16. Polymorphism
Different classes can provide the same method name with different implementations.


````python
class CSVSource:
    def read(self):
        return "Reading CSV"

class JSONSource:
    def read(self):
        return "Reading JSON"

sources = [CSVSource(), JSONSource()]

for source in sources:
    print(source.read())
````

````python
#Abstraction:
class ATM:

    def withdraw(self, amount):
        self.__check_balance(amount)
        self.__verify_pin()
        print("Please collect your cash")

    def __check_balance(self, amount):
        print("Checking account balance...")

    def __verify_pin(self):
        print("Verifying PIN...")


atm = ATM()
atm.withdraw(100)
````

## 17. Decorators
A decorator wraps a function to add behavior without rewriting the original function.


````python
def log_execution(func):
    def wrapper(*args, **kwargs):
        print(f"Starting {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Finished {func.__name__}")
        return result

    return wrapper

@log_execution
def transform_data():
    print("Transforming data...")

transform_data()
````

## 18. Generators
A generator produces values one at a time using `yield`, instead of building the entire result in memory.


````python
def batch(data, batch_size):
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]

records = list(range(1, 11))

for batch in batch(records, 3):
    print("Batch:", batch)
````

## 19. Multithreading
Threads can be useful for I/O-bound work such as waiting for APIs or files.


````python
from concurrent.futures import ThreadPoolExecutor
import time

def fetch_source(source):
    time.sleep(1)
    return f"Finished {source}"

sources = ["API-1", "API-2", "API-3"]

with ThreadPoolExecutor(max_workers=3) as executor:
    results = list(executor.map(fetch_source, sources))

print(results)
````

## 22
. Pandas Basics
Pandas uses DataFrames for tabular data processing.


````python
import pandas as pd

df = pd.DataFrame({
    "id": [1, 2, 3, 4, 4],
    "department": ["IT", "HR", "IT", "Finance", "Finance"],
    "salary": [55000, None, 65000, 58000, 58000]
})

display(df)
print(df.shape)
print(df.columns.tolist())
print(df.dtypes)
````

### Filtering, Nulls, Duplicates and Sorting


````python
clean = df.drop_duplicates().copy()
clean["salary"] = clean["salary"].fillna(0)

display(clean[clean["salary"] > 50000])
display(clean.sort_values("salary", ascending=False))
````

### groupby and Aggregation


````python
summary = (
    clean.groupby("department")
         .agg(
             average_salary=("salary", "mean"),
             employee_count=("id", "count")
         )
         .reset_index()
)

display(summary)
````

## 24. Pandas Joins & Transformations
`merge()` joins DataFrames. Other common transformations include rename, fillna, dropna and drop_duplicates.


````python
customers = pd.DataFrame({
    "customer_id": [1, 2, 3],
    "customer_name": ["John", "Sam", "Alex"]
})

transactions = pd.DataFrame({
    "transaction_id": [101, 102, 103, 104],
    "customer_id": [1, 1, 2, 3],
    "amount": [100, 150, None, 300]
})

transactions["amount"] = transactions["amount"].fillna(0)
transactions = transactions.rename(columns={"amount": "transaction_amount"})

joined = customers.merge(transactions, on="customer_id", how="left")
display(joined)
````

## 25. Mini ETL Project
Extract from CSV → clean/transform → aggregate → load to output CSV.


````python
import pandas as pd

raw = pd.DataFrame({
    "transaction_id": [1, 2, 2, 3, 4, 5],
    "customer": ["John", "Sam", "Sam", "Alex", "Maria", "Tom"],
    "city": ["London", "Coventry", "Coventry", "London", "Birmingham", "London"],
    "amount": [100, 250, 250, None, 500, 50]
})

raw.to_csv("raw_sales.csv", index=False)
display(raw)
````

````python
# EXTRACT
df = pd.read_csv("raw_sales.csv")

# TRANSFORM
df = df.drop_duplicates()
df["amount"] = df["amount"].fillna(0)
df = df[df["amount"] > 0]

city_sales = (
    df.groupby("city")
      .agg(
          total_sales=("amount", "sum"),
          transaction_count=("transaction_id", "count")
      )
      .reset_index()
)

display(city_sales)
````

````python
# LOAD
city_sales.to_csv("city_sales_summary.csv", index=False)
print("ETL completed: city_sales_summary.csv created")
````
